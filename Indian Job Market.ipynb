```python

import pandas as pd
from pathlib import Path

file_path = Path.home() / "Desktop" / "indian_tech_career_intelligence_2026 (1).xlsx"

df = pd.read_excel(file_path)

print("Data imported successfully!")
print("Dataset shape:", df.shape)
print("\nFirst 5 rows:")
print(df.head())
print("\nColumn names:")
print(df.columns.tolist())
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python

print("Rows and Columns:", df.shape)
print(df.head())
print(df.columns.tolist())
print(df.info())
```

```python

print(df.isnull().sum())
print("Total missing values:", df.isnull().sum().sum())
```

```python

df = df.drop_duplicates()
print("After removing duplicates:", df.shape)
```

```python

df.columns = df.columns.str.strip()
print(df.columns.tolist())
```

```python

# Add unique Job_ID
if "job_id" not in df.columns:
    df.insert(0, "job_id", range(1, len(df) + 1))

print(df.head())
print(df.columns.tolist())
```

```python

from pathlib import Path

desktop = Path.home() / "Desktop"

# File 1: Complete cleaned dataset
df.to_csv(desktop / "clean_jobs.csv", index=False)

print("File 1 saved successfully")
```

```python

# Create separate job skills dataset
skills_df = df[["job_id", "job_title", "skills_extracted"]].copy()

# Save second file
skills_df.to_csv(
    Path.home() / "Desktop" / "job_skills.csv",
    index=False
)

print("Second file saved successfully!")
print(skills_df.head())
```

```python

# Check missing values in each column
missing = df.isnull().sum()

print("Missing Values:")
print(missing[missing > 0])

print("Total Missing Values:", df.isnull().sum().sum())
```

```python
print(df.dtypes)
```

```python

# Convert posted_date to datetime format
df["posted_date"] = pd.to_datetime(
    df["posted_date"],
    errors="coerce"
)

# Check date format
print(df["posted_date"].dtype)
print("Invalid dates:", df["posted_date"].isna().sum())
```

```python

# Check salary data types
print(df[["salary_minimum", "salary_maximum"]].dtypes)

# Convert salary columns to numeric
df["salary_minimum"] = pd.to_numeric(
    df["salary_minimum"], errors="coerce"
)

df["salary_maximum"] = pd.to_numeric(
    df["salary_maximum"], errors="coerce"
)

print("Salary columns cleaned!")
print(df[["salary_minimum", "salary_maximum"]].dtypes)
```

```python

# Final data validation

print("Final Rows and Columns:", df.shape)

print("Duplicate Rows:", df.duplicated().sum())

print("Missing Values:")
print(df.isnull().sum())

print("Data Types:")
print(df.dtypes)

print("First 5 Rows:")
print(df.head())
```

```python

# Final cleaning verification
print("Dataset Shape:", df.shape)
print("Duplicate Rows:", df.duplicated().sum())
print("Missing Values:", df.isnull().sum().sum())
print("Unique Job IDs:", df["job_id"].nunique())

print("\nCleaning Verification Completed!")
```

```python

# Top 10 Job Titles
top_jobs = df["job_title"].value_counts().head(10)

print("Top 10 Job Titles:")
print(top_jobs)
```

```python

# Experience level analysis
print("Job Count by Experience Level:")
print(df["experience_level"].value_counts())

# Experience required analysis
print("\nExperience Required:")
print(df["experience_required"].value_counts())
```

```python

# Work mode analysis
print("Job Count by Work Mode:")
print(df["work_mode"].value_counts())

# Percentage distribution
print("\nWork Mode Percentage:")
print(
    (df["work_mode"].value_counts(normalize=True) * 100)
    .round(2)
)
```

```python

# Salary analysis
print("Minimum Salary Statistics:")
print(df["salary_minimum"].describe())

print("\nMaximum Salary Statistics:")
print(df["salary_maximum"].describe())

# Average salary range
df["average_salary"] = (
    df["salary_minimum"] + df["salary_maximum"]
) / 2

print("\nAverage Salary:")
print(df["average_salary"].mean())
```

```python

from collections import Counter

# Combine all skill values
all_skills = df["skills_extracted"].dropna().astype(str)

# Split skills and count frequency
skill_counts = Counter()

for skills in all_skills:
    for skill in skills.replace(",", ";").split(";"):
        skill = skill.strip()
        if skill:
            skill_counts[skill] += 1

# Display top 15 skills
print("Top 15 Most Demanded Skills:")
for skill, count in skill_counts.most_common(15):
    print(skill, ":", count)
```

```python

# City-wise job analysis
print("Top 10 Cities with Most Job Postings:")

print(
    df["location_city"]
    .value_counts()
    .head(10)
)
```

```python

# Industry-wise job analysis
print("Top 10 Industries by Job Postings:")

print(
    df["industry"]
    .value_counts()
    .head(10)
)
```

```python

print("Average Salary by Experience Level:")

print(
    df.groupby("experience_level")["average_salary"]
    .mean()
    .sort_values(ascending=False)
)
```

```python


print("Work Mode by Experience Level:")

print(pd.crosstab(df["experience_level"], df["work_mode"]))

```

```python

print("Jobs by Posting Year:")

print(df["posted_date"].dt.year.value_counts().sort_index())
```

```python

output = Path.home() / "Desktop" / "job_market_analysis.xlsx"

with pd.ExcelWriter(output) as writer:
    df.to_excel(writer, sheet_name="Cleaned_Data", index=False)
    df["job_title"].value_counts().head(10).to_excel(
        writer, sheet_name="Top_Job_Titles"
    )
    df["location_city"].value_counts().head(10).to_excel(
        writer, sheet_name="Top_Cities"
    )

print("Analysis results saved!")
```
